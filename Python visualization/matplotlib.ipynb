```python
import matplotlib.pyplot as plt
import pandas as pd 
import numpy as np

%matplotlib inline

plt.figure()
plt.plot()
plt.show()
```

***BAR CHART***

```python
x=["python","java","c++","c"]
y=[85,80,65,72]
plt.bar(x,y)
plt.show()
```

BAR CHART WITH ATTRIBUTES


```python
x=["python","java","c++","c"]
y=[85,80,65,72]
c=["y","m","g","b"]
plt.xlabel("PROGRAMMING LANGUAGES",fontsize=15)
plt.ylabel("POPULARITY",fontsize=15)
plt.title("POPULARITY AMONG PROGRAMMING LANGUAGES",fontsize=20)
plt.bar(x,y,color=c,width=0.4,align="edge",edgecolor='r',linewidth=5,alpha=0.5,linestyle=":",label="DEVA")
plt.legend()
plt.show()


#aligns --> edge,centre
#width -->bar size
#edgecolor --> bar border color
#linewidth --> bar boundary linesize
#alpha --> used to blur the bar , it ranges 1-0
#linestyle --> bar boundary style 
```

****DOUBLE BAR CHART****

```python
x=["python","java","c++","c"]
y=[85,80,65,72]
z=[60,85,55,32]


width=0.3


p=np.arange(len(x))
p1=[j+width for j in p]

plt.xlabel("PROGRAMMING LANGUAGES",fontsize=15)
plt.ylabel("POPULARITY",fontsize=15)
plt.title("POPULARITY AMONG PROGRAMMING LANGUAGES",fontsize=20)

plt.bar(p,y,width,color="r",label="popularity 2023")
plt.bar(p1,z,width,color="y",label="popularity 2024")


plt.xticks(p+width/2,x,rotation=10)
plt.legend()
plt.show()
```

HORIZONTAL BAR GRAPH

```python
x=["python","java","c++","c"]
y=[85,80,75,72]
z=[60,85,55,32]

plt.xlabel("PROGRAMMING LANGUAGES",fontsize=15)
plt.ylabel("POPULARITY",fontsize=15)
plt.title("POPULARITY AMONG PROGRAMMING LANGUAGES",fontsize=20)
plt.barh(x,y,color="r",label="popularity 1")
plt.barh(x,z,color="y",label="popularity 2")
plt.legend()
plt.show()
```

SCATTER CHART

```python
#simple scatter plot
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
plt.scatter(day,no)
plt.show()
```

```python
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
plt.xlabel("day",fontsize=15)
plt.ylabel("no",fontsize=15)
plt.title("Scatter plot",fontsize=15)
colors=["r","g","y","b","y","r","m"]
sizes=[400,200,300,200,500,300,100]
plt.scatter(day,no,c=colors,s=sizes,alpha=0.4,marker="*",edgecolor="g",linewidth=1)
plt.show()

#we can give multiple colors also by {c} parameter
#we can give different sizes by {s} parameter
#alpha -->blur
#marker will be used to change the scatter shape
#edgecolor-->marker edge
#linewidth -->marker line size

```

SCATTER USING COLOR BARS

```python
#colors
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
plt.xlabel("day",fontsize=15)
plt.ylabel("no",fontsize=15)
plt.title("Scatter plot",fontsize=15)
colors=[10,49,30,29,56,20,30]                                                       #colors range 0 to 100
sizes=[400,200,300,200,500,300,100]
plt.scatter(day,no,c=colors,s=sizes,cmap="bwr")                                      #to plot color bar
t=plt.colorbar()
t.set_label("colorbar",fontsize=15)
plt.show()
```

MULTIPLE SCATTER PLOTS

```python
#MULTIPLE SCATTER PLOT
day=[1,2,3,4,5,6,7]
no=[2,3,1,4,5,3,6]
no2=[3,2,4,5,1,6,2]
plt.xlabel("day",fontsize=15)
plt.ylabel("no",fontsize=15)
plt.title("Scatter plot",fontsize=15)
colors=[10,49,30,29,56,20,30]                                                       #colors range 0 to 100
sizes=[400,200,300,200,500,300,100]
plt.scatter(day,no,c=colors,s=sizes,cmap="viridis",alpha=0.5)
plt.scatter(day,no2,color="r")                                     
t=plt.colorbar()
t.set_label("colorbar",fontsize=15)
plt.show()
```

```python
no=[41, 54, 52, 59, 38, 34, 25, 24, 52, 46, 46, 35, 23, 19, 44, 22, 33,
       48, 11, 42, 17, 27, 37, 11, 18, 56, 16, 59, 16, 30, 52, 35, 46, 32,
       41, 45, 53, 15, 17, 30, 18, 11, 11, 45, 19, 15, 53, 45, 12, 31]
plt.hist(no)
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()    
```

```python
#parameters
plt.hist(no,color="g",edgecolor="r")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   
```

```python
#bins
l=[10,20,30,40,50,60]
plt.hist(no,color="g",bins=l,edgecolor="r")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   
```

```python
#to give more range
plt.hist(no,"auto",(0,200),edgecolor="r")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   
```

```python
#cumulative will give reverse graph
plt.hist(no,color="g",edgecolor="r",cumulative=-1)
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   
```

```python
plt.hist(no,color="g",edgecolor="r",bottom=10,align="left")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   


#bottom -->to change bottom value
#align --> left/mid
```

```python
#hist types
plt.hist(no,color="g",edgecolor="r",histtype="step")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   

#step,stepfilledbar,barstacked
```

```python
#hist types
plt.hist(no,color="g",edgecolor="r",histtype="barstacked",orientation="horizontal")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   
```

```python
#axvline -->used to point important line for discussion
plt.hist(no,color="m",bins=1,edgecolor="r",label="hist")

plt.axvline(45,color="g",label="line")
plt.ylabel("y")
plt.xlabel("x")
plt.title("histo")
plt.show()   
```

PIE CHART

```python
x=[10,20,30,40]
y=["c","c++","java","python"]

plt.pie(x)

plt.show()
```

```python
#parameters
x=[10,20,30,40]
y=["c","c++","java","python"]
ex=[0.4,0.0,0.0,0.0]
c=["r","m","g","y"]

plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.1f%%",shadow=True,radius=1.3,labeldistance=1.2,startangle=90,textprops={"fontsize":15},
        counterclock=False,wedgeprops={"linewidth":1,"edgecolor":"b"},rotatelabels=True)


plt.title("pie chart")



plt.show()

#labels --> names
#explode --> to seperate pie
#colors --> to give multiple colors
#autopct --> to give percentages
#shadow --> to create shadow
#radius --> to change size of circle
#labeldistance--> to change the distance of label
#startangle--> to decide the starting angle --> o is default
#textprops --> text properties(fontsize)
#counterclock--> to change direction(true/false)
#wedgeprops -->to determine border length and color
#rotatelabels --> it will rotate the labels
```
